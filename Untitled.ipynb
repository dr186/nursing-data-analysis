```python
# 1. 手动造一个10行的模拟护士排班表
data = {
    "姓名": ["张三", "李四", "王五", "张三", "李四", "王五", "张三", "李四", "王五", "张三"],
    "班次": ["白班", "夜班", "夜班", "夜班", "白班", "白班", "夜班", "夜班", "白班", "白班"],
    "科室": ["ICU", "ICU", "急诊", "急诊", "ICU", "急诊", "ICU", "急诊", "ICU", "急诊"]
}

# 2. 把字典转换成 pandas 的表格格式
df = pd.DataFrame(data)

# 3. 显示这个表格
df
```

```python
#1.先把所有"夜班"的行单独挑出来看看
df[df["班次"]=="夜班"]
```

```python
#2.统计每个人分别上了几次夜班
df.groupby("姓名")["班次"].apply(lambda x:(x=="夜班").sum())
```

```python
#按科室统计，每个科室一共有多少个夜班
df.groupby("科室")["班次"].apply(lambda x:(x=="夜班").sum())
```

```python
#把刚才统计的“每个人夜班次数”保存为Excel
result=df.groupby("姓名")["班次"].apply(lambda x:(x=="夜班").sum())
result.to_excel("夜班统计表.xslx")
print("Excel文件已生成")
                                        
```
