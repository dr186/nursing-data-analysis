```python
import pandas as pd
import numpy as np

# 模拟真实排班数据
data = {
    "日期": ["2026-11-01", "2026-11-01", "2026-11-01", "2026-11-02", "2026-11-02", "2026-11-02", "2026-11-03", "2026-11-03", "2026-11-03"],
    "科室": ["ICU", "ICU", "急诊", "急诊", "ICU", "急诊", "ICU", "急诊", "ICU"],
    "姓名": ["张三", "李四", "王五", "张三", "李四", "王五", "张三", "赵六", "李四"],
    "班次": ["夜班", "白班", "夜班", "夜班", "夜班", "白班", "夜班", "夜班", "白班"]
}
df = pd.DataFrame(data)

# 把日期转换成 datetime 格式（昨天学过的）
df["日期"] = pd.to_datetime(df["日期"])
print("数据已生成！")
df
```

```python
import pandas as pd
# 使用crosstab函数，行是科室，列是班次，统计次数
report_keshi = pd.crosstab(index=df["科室"],columns=df["班次"])
print("=== 科室排版统计表 ===")
print(report_keshi)
```

```python
print(report_keshi)
print("=== 科室排班统计表 ===")
```

```python
import pandas as pd
df_sorted = df.sort_values(by=["姓名","日期"])
df
```

```python
df_sorted["前一天的班次"] = df_sorted.groupby("姓名")["班次"].shift(1)
df_sorted["前一天的日期"] = df_sorted.groupby("姓名")["日期"].shift(1)
condition = ((df_sorted["班次"] == "夜班")&
            (df_sorted["前一天的班次"] == "夜班")&
            ((df_sorted["日期"] - df_sorted["前一天的日期"]).dt.days == 1))
violations = df_sorted[condition][["姓名","日期"]]
violations["违规原因"] = "连续两天夜班"
print("=== 违规排班名单 ===")
print(violations)
             
```

```python
import pandas as pd

# ================= 第一关：生成数据 =================
data = {
    "日期": ["2026-11-01", "2026-11-01", "2026-11-01", "2026-11-02", "2026-11-02", "2026-11-02", "2026-11-03", "2026-11-03", "2026-11-03"],
    "科室": ["ICU", "ICU", "急诊", "急诊", "ICU", "急诊", "ICU", "急诊", "ICU"],
    "姓名": ["张三", "李四", "王五", "张三", "李四", "王五", "张三", "赵六", "李四"],
    "班次": ["夜班", "白班", "夜班", "夜班", "夜班", "白班", "夜班", "夜班", "白班"]
}
df = pd.DataFrame(data)
df["日期"] = pd.to_datetime(df["日期"])

# ================= 第二关：科室统计 =================
report_keshi = pd.crosstab(index=df["科室"], columns=df["班次"])

# ================= 第三关：抓出违规名单 =================
df_sorted = df.sort_values(by=["姓名", "日期"])
df_sorted["前一天的班次"] = df_sorted.groupby("姓名")["班次"].shift(1)
df_sorted["前一天的日期"] = df_sorted.groupby("姓名")["日期"].shift(1)

condition = (
    (df_sorted["班次"] == "夜班") & 
    (df_sorted["前一天的班次"] == "夜班") & 
    ((df_sorted["日期"] - df_sorted["前一天的日期"]).dt.days == 1)
)
violations = df_sorted[condition][["姓名", "日期"]].copy()
violations["违规原因"] = "连续两天夜班"

# ================= 第四关：一键导出多Sheet =================
file_name = "11月质控报表.xlsx"
with pd.ExcelWriter(file_name, engine="openpyxl") as writer:
    df.to_excel(writer, sheet_name="原始排班", index=False)
    report_keshi.to_excel(writer, sheet_name="科室统计", index=True)
    violations.to_excel(writer, sheet_name="违规名单", index=False)

print(f"✅ 大功告成！请在左侧文件列表双击打开：{file_name}")
```
