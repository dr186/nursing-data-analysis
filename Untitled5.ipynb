```python
import pandas as pd

# 模拟一份包含了 日期、姓名、班次 的排班表
data = {
    "日期": ["2026-11-01", "2026-11-01", "2026-11-02", "2026-11-02", "2026-11-03", "2026-11-03", "2026-11-04", "2026-11-04"],
    "姓名": ["张三", "李四", "张三", "王五", "李四", "赵六", "王五", "张三"],
    "班次": ["白班", "夜班", "夜班", "白班", "夜班", "请假", "夜班", "夜班"]
}
df_real = pd.DataFrame(data)

# 注意看张三：11月2日上了夜班，11月3日没排他，11月4日又上了夜班。
# 实际上，医院是不允许夜班后第二天直接接着上白班或夜班的，这叫“连班违规”。

df_real.to_excel("真实排班_带冲突.xlsx", index=False)
print("真实复杂表格已生成！")
```
