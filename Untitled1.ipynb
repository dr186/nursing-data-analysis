```python
import warnings
warnings.filterwarnings("ignore") # 屏蔽烦人的红字警告

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import font_manager
import shutil
import os

# 1. 自动清理旧的字体缓存
cache_dir = os.path.expanduser('~/.cache/matplotlib')
if os.path.exists(cache_dir):
    shutil.rmtree(cache_dir)
    print("✅ 字体缓存已清理，下次画图会重新加载字体")

# 2. 自动扫描系统里有哪些中文字体
font_names = [f.name for f in font_manager.fontManager.ttflist]
chinese_fonts = [f for f in font_names if 'WenQuanYi' in f or 'Noto' in f or 'Hei' in f]
print(f"🔍 系统找到的中文字体有: {chinese_fonts[:3]}")

# 3. 强行指定第一个找到的中文字体
if chinese_fonts:
    plt.rcParams['font.sans-serif'] = [chinese_fonts[0]]
    plt.rcParams['axes.unicode_minus'] = False

# 4. 重新读取和清洗数据
df = pd.read_excel("排班表2.xlsx")
df["姓名"] = df["姓名"].str.strip()
df_clean = df.dropna()
counts = df_clean[df_clean["班次"] == "夜班"]["姓名"].value_counts()

# 5. 画图
plt.figure(figsize=(8, 5))
plt.bar(counts.index, counts.values, color='skyblue')
plt.title("各护士夜班次数统计")
plt.xlabel("姓名")
plt.ylabel("夜班次数")
plt.show()
```
