```python
import pandas as pd
import numpy as np

# 设定随机种子，保证每次生成的数据一样，方便核对
np.random.seed(42)

# 模拟 11 月份的 30 天
dates = pd.date_range(start="2026-11-01", end="2026-11-30")
names = ["张三", "李四", "王五", "赵六", "孙七"]
depts = ["ICU", "急诊", "呼吸科"]
shifts = ["白班", "夜班", "休息"]

# 生成随机数据：随机分配科室，随机分配班次
data = []
for date in dates:
    for name in names:
        data.append({
            "日期": date,
            "姓名": name,
            "科室": np.random.choice(depts),
            "班次": np.random.choice(shifts, p=[0.5, 0.3, 0.2]) # 50%白班，30%夜班，20%休息
        })

df_dashboard = pd.DataFrame(data)
df_dashboard.to_excel("11月整月排班_大数据.xlsx", index=False)
print("✅ 整月数据已生成，共", len(df_dashboard), "条记录！")
```

```python
import warnings
warnings.filterwarnings("ignore") # 忽略所有警告
import matplotlib.pyplot as plt

# Windows 下的中文字体设置（黑体）
plt.rcParams['font.sans-serif'] = ['WenQuanYi Zen Hei']
plt.rcParams['axes.unicode_minus'] = False

# 1. 筛选出所有夜班记录
df_night = df_dashboard[df_dashboard["班次"] == "夜班"]

# 2. 按“日期”分组，统计每天有多少个夜班
daily_nights = df_night.groupby("日期").size()

# 3. 画折线图
plt.figure(figsize=(12, 5)) # 把画布拉宽一点，好看清30天
plt.plot(daily_nights.index, daily_nights.values, marker='o', color='#FF7F50', linewidth=2)

plt.title("11月每日夜班数量趋势分析", fontsize=16)
plt.xlabel("日期", fontsize=12)
plt.ylabel("夜班人数", fontsize=12)
# 把X轴的日期斜着放，防止字叠在一起
plt.xticks(rotation=45) 
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout() # 自动调整布局防遮挡

# 保存图片
plt.savefig("11月夜班趋势折线图.png", dpi=300)
plt.show()
```

```python
# 1. 生成科室和班次的交叉表（统计次数）
crosstab = pd.crosstab(df_dashboard["科室"], df_dashboard["班次"])

# 2. 把“休息”这一列删掉，我们只看工作量
crosstab_work = crosstab.drop(columns=["休息"])

print("各科室白班夜班统计：")
print(crosstab_work)

# 3. 直接使用 pandas 自带的 plot 画分组柱状图
ax = crosstab_work.plot(kind='bar', figsize=(10, 6), color=['#87CEEB', '#FF7F50'], edgecolor='black')

plt.title("各科室白班与夜班工作量对比", fontsize=16)
plt.xlabel("科室", fontsize=12)
plt.ylabel("班次数量 (个)", fontsize=12)
plt.xticks(rotation=0) # X轴科室名字水平放置
plt.legend(["白班", "夜班"]) # 设置图例

# 关键一步：在柱子上把具体的数字写出来，显得极其专业！
for p in ax.patches:
    ax.annotate(str(p.get_height()), (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.tight_layout()
plt.savefig("科室工作量对比柱状图.png", dpi=300)
plt.show()
```
