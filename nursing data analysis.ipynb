```python
import pandas as pd
# 1.读取你之前传上去的Excel文件
df=pd.read_excel("夜班统计表.xlsx")
# 2.看看数据长什么样
df
```

```python
#故意把第一行的"班次"变成空值(模拟护士填漏了)
df.loc[0,"班次"]=None
#查看缺失值
print("清理前的缺失值:\n",df.isna().sum())
#删掉有缺失值的行
df_clean=df.dropna()
#重新统计张三的夜班数
print("\n清理后的结果:")
df_clean.groupby("姓名")["班次"].apply(lambda x:(x=="夜班").sum())
```
