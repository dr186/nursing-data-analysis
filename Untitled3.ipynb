```python
%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd

# 1. 设定 Windows 下的中文字体（Windows 用黑体）
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# 2. 准备假数据（科室班次构成）
labels = ['白班', '夜班', '下夜班休息', '请假']
sizes = [45, 30, 20, 5]  
colors = ['#87CEEB', '#FF7F50', '#90EE90', '#D3D3D3'] 

# 3. 画饼图
plt.figure(figsize=(7, 7))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90, shadow=True)
plt.title("某科室班次类型占比", fontsize=16)
plt.show()

# 4. 把图保存成高清 PNG 图片，存到当前文件夹
plt.savefig("班次占比饼图.png", dpi=300, bbox_inches='tight')
print("✅ 高清图片已保存！")
```
